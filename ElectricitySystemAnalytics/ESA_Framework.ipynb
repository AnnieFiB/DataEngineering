# Electricity System Analytics Framework


## 1. Purpose
Provide a structured, repeatable method for analysing electricity system performance across:
- Decarbonisation
- Security & dependency
- Flexibility & system behaviour
- Regional equity

All metrics are derived from the Electricity Maps API.

---

## 2. Data Sources (Electricity Maps API)
- `/v3/carbon-intensity/past` – historical carbon intensity  
- `/v3/carbon-intensity/latest` – real-time carbon intensity  
- `/v3/generation-mix/past` – historical generation by fuel  
- `/v3/exchanges/past` – historical imports/exports  
- `/v3/zones` – list of zones and identifiers  

---

## 3. Core KPIs

### 3.1 Average Carbon Intensity (Operational)
**Formula:**  
Avg CI = Σ(MWhₜ × CIₜ) / Σ(MWhₜ)

**API Inputs:**  
- Carbon intensity (hourly)  
- Total generation (from generation mix)

---

### 3.2 Renewable Penetration
**Formula:**  
Renewable Share = Renewable MWh / Total MWh

**API Inputs:**  
- Generation mix by fuel type

---

### 3.3 Fossil Dependency Ratio
**Formula:**  
Fossil Share = Fossil MWh / Total MWh

---

### 3.4 Import Dependency
**Formula:**  
Import Share = Net Imports / (Generation + Imports)

**API Inputs:**  
- `/v3/exchanges/past`

---

### 3.5 Imported Carbon Share
**Formula:**  
Imported CO₂ = Σ(Imported MWh × Exporter CI)

---

### 3.6 Carbon Volatility Index
**Formula:**  
StdDev of hourly carbon intensity over period

---

### 3.7 Low‑Carbon Hours Share
**Formula:**  
Hours with CI < threshold / Total hours

---

## 4. Analytical Pillars

### 4.1 Decarbonisation
- Trend in CI  
- Renewable penetration  
- Fossil dependency  

### 4.2 Security & Dependency
- Import share  
- Imported carbon intensity  
- Exposure to external shocks  

### 4.3 Flexibility & System Behaviour
- CI volatility  
- Renewable variability  
- Low‑carbon windows  

### 4.4 Regional Equity
- Regional CI  
- Regional renewable share  
- Regional dependency
